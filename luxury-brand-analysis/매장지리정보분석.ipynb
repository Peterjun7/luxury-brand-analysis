```python
#지도시각화
import pandas as pd
import numpy as np
import folium
```

```python
#hermes
hermes_geo = pd.read_csv('hermes_geo.shp.csv', encoding = 'cp949', engine = 'python')
hermes_geo.head()
```

```python
map_hermes = folium.Map(location = [36.7219941, 127.4959787], tiles = 'cartodbpositron', zoom_start = 6)
```

```python
for i, store in hermes_geo.iterrows() :
    folium.Marker(location = [store['위도'], store['경도']],
                  tooltip = store['store'],
                  icon = folium.Icon(color = 'orange', icon = 'gift', prefix = 'fa')).add_to(map_hermes)

map_hermes.save('map_hermes.html')    
map_hermes
```
