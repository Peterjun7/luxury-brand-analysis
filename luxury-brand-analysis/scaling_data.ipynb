```python
#최종분석
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import spatial
from scipy.spatial import distance

df = pd.DataFrame([[14851224, 672, 3.6], [3253674, 652, 2.0], [6255129, 569, 3.5], [4704726, 723, 2.8]])
df.columns = ['price', 'affection', 'density']
x = ['Hermes', 'Louis Vuitton', 'Chanel', 'Christian Dior']
```

```python
df['price_minmax'] = (df['price'] - df['price'].min()) / (df['price'].max() - df['price'].min())
df['affection_minmax'] = (df['affection'] - df['affection'].min()) / (df['affection'].max() - df['affection'].min())
df['density_minmax'] = (df['density'] - df['density'].min()) / (df['density'].max() - df['density'].min())
df.head()
```

```python
plt.title('Price_min max')
plt.bar(x, df['price_minmax'], width = 0.5, color = 'darkgreen')
plt.legend(df.columns[0])
plt.savefig('price_minmax.png', dpi=200)
```

```python
plt.title('Affection_min max')
plt.bar(x, df['affection_minmax'], width = 0.5, color = 'darkred')
plt.legend(df.columns[1])
plt.savefig('affection_minmax.png', dpi=200)
```

```python
plt.title('Density_min max')
plt.bar(x, df['density_minmax'], width = 0.5, color = 'darkblue')
plt.legend(df.columns[2])
plt.savefig('density_minmax.png', dpi=200)
```

```python
df['sum_minmax'] = df['price_minmax'] + df['affection_minmax'] + df['density_minmax']
```

```python
plt.title('Result_min max')
plt.bar(x, df['sum_minmax'], width = 0.5, color = 'gold')
plt.legend('sum')
plt.savefig('sum_minmax.png', dpi=200)
```

```python
df['price_log'] = np.log1p(df['price'])
df['affection_log'] = np.log1p(df['affection'])
df['density_log'] = np.log1p(df['density'])
df.head()
```

```python
plt.title('Price_log')
plt.plot(x, df['price_log'], color = 'olivedrab', marker = 'o', linestyle = '--')
plt.xlabel('Brands')
plt.ylabel('Price')
plt.savefig('price_log.png', dpi=200)
```

```python
plt.title('Affection_log')
plt.plot(x, df['affection_log'], color = 'crimson', marker = 'o', linestyle = '--')
plt.xlabel('Brands')
plt.ylabel('Affection')
plt.savefig('affection_log.png', dpi=200)
```

```python
plt.title('Density_log')
plt.plot(x, df['density_log'], color = 'steelblue', marker = 'o', linestyle = '--')
plt.xlabel('Brands')
plt.ylabel('Density')
plt.savefig('density_log.png', dpi=200)
```

```python
df['sum_log'] = df['price_log'] + df['affection_log'] + df['density_log']
```

```python
plt.title('Result')
plt.plot(x, df['sum_log'], color = 'gold', marker = 'o', linestyle = '--')
plt.xlabel('Brands')
plt.ylabel('Sum')
plt.savefig('sum_log.png', dpi=200)
```
